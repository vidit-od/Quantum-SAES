# Efficient PRINCE Circuit

This notebook builds the same PRINCE computation as `Main.ipynb`, but removes full-circuit barriers from the round and S-layer construction so independent S-boxes can be scheduled in parallel. The raw circuit uses only Clifford gates (`x`, `cx`) and Toffoli (`ccx`).

```python
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister, transpile
from qiskit.circuit.library import RCCXGate
from qiskit_aer import AerSimulator

from Helpers import xor_constant_into_register, xor_register_into_register
from Sbox import (
    quantum_sbox_nibble,
    quantum_sbox_inverse_nibble,
    classical_s_layer,
    classical_s_layer_inv,
)
from Mbox import quantum_m_layer, quantum_m_layer_inv, classical_m_layer, classical_m_layer_inv
from Round import classical_round, classical_inverse_round
```

```python
ROUND_CONSTANTS = [
    0x0000000000000000,
    0x13198A2E03707344,
    0xA4093822299F31D0,
    0x082EFA98EC4E6C89,
    0x452821E638D01377,
    0xBE5466CF34E90C6C,
    0x7EF84F78FD955CB1,
    0x85840851F1AC43AA,
    0xC882D32F25323C54,
    0x64A51195E0E3610D,
    0xD3B5A399CA0C2399,
    0xC0AC29B7C97C50DD,
]

MASK64 = (1 << 64) - 1
RC0 = ROUND_CONSTANTS[0]
ALPHA = ROUND_CONSTANTS[11]


def prince_key_schedule(K_128bit):
    k0 = (K_128bit >> 64) & MASK64
    k1 = K_128bit & MASK64
    ror1 = ((k0 >> 1) | (k0 << 63)) & MASK64
    k0_prime = ror1 ^ (k0 >> 63)
    return k0, k1, k0_prime
```

## No-Barrier Quantum Core

The important optimization is here: the 16 S-boxes share no qubits, so we append them without barriers. Qiskit can then schedule them into the same depth layers.

```python
def efficient_s_layer(qc, state, anc):
    for i in range(16):
        nibble = [state[4 * i + k] for k in range(4)]
        scratch = [anc[14 * i + k] for k in range(14)]
        quantum_sbox_nibble(qc, nibble, scratch)


def efficient_s_layer_inv(qc, state, anc):
    for i in range(16):
        nibble = [state[4 * i + k] for k in range(4)]
        scratch = [anc[14 * i + k] for k in range(14)]
        quantum_sbox_inverse_nibble(qc, nibble, scratch)


def efficient_round(qc, state, anc, qcore, rc):
    efficient_s_layer(qc, state, anc)
    quantum_m_layer(qc, state)
    xor_constant_into_register(qc, state, rc)
    xor_register_into_register(qc, qcore, state)


def efficient_inverse_round(qc, state, anc, qcore, rc):
    xor_register_into_register(qc, qcore, state)
    xor_constant_into_register(qc, state, rc)
    quantum_m_layer_inv(qc, state)
    efficient_s_layer_inv(qc, state, anc)
```

```python
def build_efficient_prince_circuit(
    input_64bit=0,
    K_128bit=0,
    mode="encrypt",
    include_initialization=False,
    include_measurement=False,
):
    k0, k1, k0_prime = prince_key_schedule(K_128bit)
    if mode == "encrypt":
        pre_key, core_key, post_key = k0, k1, k0_prime
    elif mode == "decrypt":
        pre_key, core_key, post_key = k0_prime, k1 ^ ALPHA, k0
    else:
        raise ValueError("mode must be 'encrypt' or 'decrypt'")

    state = QuantumRegister(64, name="state")
    anc = QuantumRegister(16 * 14, name="anc")
    qpre = QuantumRegister(64, name="pre_key")
    qcore = QuantumRegister(64, name="core_key")
    qpost = QuantumRegister(64, name="post_key")
    registers = [state, anc, qpre, qcore, qpost]

    creg = None
    if include_measurement:
        creg = ClassicalRegister(64, name="out")
        registers.append(creg)

    qc = QuantumCircuit(*registers)

    if include_initialization:
        for bit in range(64):
            if (input_64bit >> bit) & 1:
                qc.x(state[bit])
            if (pre_key >> bit) & 1:
                qc.x(qpre[bit])
            if (core_key >> bit) & 1:
                qc.x(qcore[bit])
            if (post_key >> bit) & 1:
                qc.x(qpost[bit])

    xor_register_into_register(qc, qpre, state)
    xor_register_into_register(qc, qcore, state)
    xor_constant_into_register(qc, state, RC0)

    for i in range(1, 6):
        efficient_round(qc, state, anc, qcore, ROUND_CONSTANTS[i])

    efficient_s_layer(qc, state, anc)
    quantum_m_layer(qc, state)
    efficient_s_layer_inv(qc, state, anc)

    for i in range(6, 11):
        efficient_inverse_round(qc, state, anc, qcore, ROUND_CONSTANTS[i])

    xor_constant_into_register(qc, state, ROUND_CONSTANTS[11])
    xor_register_into_register(qc, qcore, state)
    xor_register_into_register(qc, qpost, state)

    if include_measurement:
        qc.measure(state, creg)

    return qc
```

## Classical References

```python
def classical_prince_with_keys(input_64bit, pre_key, core_key, post_key):
    state = input_64bit & MASK64
    state ^= pre_key
    state ^= core_key
    state ^= RC0

    for i in range(1, 6):
        state = classical_round(i, state, core_key, ROUND_CONSTANTS[i])

    state = classical_s_layer(state)
    state = classical_m_layer(state)
    state = classical_s_layer_inv(state)

    for i in range(6, 11):
        state = classical_inverse_round(i, state, core_key, ROUND_CONSTANTS[i])

    state ^= ROUND_CONSTANTS[11]
    state ^= core_key
    state ^= post_key
    return state & MASK64


def classical_encrypt(plaintext_64bit, K_128bit):
    k0, k1, k0_prime = prince_key_schedule(K_128bit)
    return classical_prince_with_keys(plaintext_64bit, k0, k1, k0_prime)


def classical_decrypt(ciphertext_64bit, K_128bit):
    k0, k1, k0_prime = prince_key_schedule(K_128bit)
    return classical_prince_with_keys(ciphertext_64bit, k0_prime, k1 ^ ALPHA, k0)
```

## Resource Estimation

```python
IGNORED_OPS = {"barrier", "measure"}
RAW_BASIS = ["x", "cx", "ccx"]
CLIFFORD_T_BASIS = ["h", "x", "z", "s", "sdg", "cx", "t", "tdg"]
CLIFFORD_OPS = {"h", "x", "z", "s", "sdg", "cx"}
T_OPS = {"t", "tdg"}


def clean_counts(qc):
    return {k: v for k, v in dict(qc.count_ops()).items() if k not in IGNORED_OPS}


def clean_depth(qc):
    return qc.depth(filter_function=lambda inst: inst.operation.name not in IGNORED_OPS)


def raw_resource_table(qc):
    optimized = transpile(qc, basis_gates=RAW_BASIS, optimization_level=3)
    ops = clean_counts(optimized)
    return {
        "total_qubits": optimized.num_qubits,
        "x_gates": ops.get("x", 0),
        "cx_gates": ops.get("cx", 0),
        "toffoli_gates": ops.get("ccx", 0),
        "total_gates": sum(ops.values()),
        "logical_depth": clean_depth(optimized),
    }


def clifford_t_resource_table(qc):
    converted = transpile(qc, basis_gates=CLIFFORD_T_BASIS, optimization_level=3)
    ops = clean_counts(converted)
    return {
        "total_qubits": converted.num_qubits,
        "total_clifford_gates": sum(ops.get(name, 0) for name in CLIFFORD_OPS),
        "total_t_gates": sum(ops.get(name, 0) for name in T_OPS),
        "t_depth": converted.depth(filter_function=lambda inst: inst.operation.name in T_OPS),
        "total_depth": clean_depth(converted),
        "gate_wise_counts": ops,
    }


efficient_qc = build_efficient_prince_circuit()
raw_resources = raw_resource_table(efficient_qc)
clifford_t_resources = clifford_t_resource_table(efficient_qc)

print("Raw Clifford+Toffoli:", raw_resources)
print("Clifford+T:", {k: v for k, v in clifford_t_resources.items() if k != "gate_wise_counts"})
print("Clifford+T gate-wise:", clifford_t_resources["gate_wise_counts"])
```

## Aggressive Relative-Phase Option

This optional pass replaces every exact Toffoli (`ccx`) with Qiskit's relative-phase Toffoli (`rccx`) before decomposing to Clifford+T. It dramatically reduces T-depth and still passes the classical-basis encryption/decryption tests below. For Grover-oracle use, this needs extra care because relative phases can matter for interference; use this table only after checking that phases are either harmless or canceled in the oracle construction.

```python
def replace_toffoli_with_relative_phase(qc):
    converted = QuantumCircuit(*qc.qregs, *qc.cregs)
    for circuit_instruction in qc.data:
        op = circuit_instruction.operation
        if op.name == "ccx":
            converted.append(RCCXGate(), circuit_instruction.qubits, circuit_instruction.clbits)
        else:
            converted.append(op, circuit_instruction.qubits, circuit_instruction.clbits)
    return converted


relative_phase_qc = replace_toffoli_with_relative_phase(efficient_qc)
relative_phase_resources = clifford_t_resource_table(relative_phase_qc)
print("Relative-phase Clifford+T:", {k: v for k, v in relative_phase_resources.items() if k != "gate_wise_counts"})
print("Relative-phase gate-wise:", relative_phase_resources["gate_wise_counts"])
```

## Main-Style Functional Tests

```python
def simulate_output(qc):
    sim = AerSimulator(method="matrix_product_state")
    result = sim.run(qc, shots=1).result()
    return int(next(iter(result.get_counts())), 2)


def maybe_prepare_for_simulation(qc, convert_to_clifford_t=False, relative_phase=False):
    if relative_phase:
        qc = replace_toffoli_with_relative_phase(qc)
    if convert_to_clifford_t or relative_phase:
        qc = transpile(qc, basis_gates=CLIFFORD_T_BASIS, optimization_level=3)
    return qc


def quantum_encrypt(input_64bit, K_128bit, convert_to_clifford_t=False, relative_phase=False):
    qc = build_efficient_prince_circuit(input_64bit, K_128bit, "encrypt", True, True)
    qc = maybe_prepare_for_simulation(qc, convert_to_clifford_t, relative_phase)
    return simulate_output(qc)


def quantum_decrypt(input_64bit, K_128bit, convert_to_clifford_t=False, relative_phase=False):
    qc = build_efficient_prince_circuit(input_64bit, K_128bit, "decrypt", True, True)
    qc = maybe_prepare_for_simulation(qc, convert_to_clifford_t, relative_phase)
    return simulate_output(qc)


def check_equal(label, got, expected):
    ok = got == expected
    print(f"{label}: 0x{got:016X}  {'PASS' if ok else 'FAIL'}")
    assert ok


# Test 1: all-zero plaintext and key.
pt1 = 0x0000000000000000
K1 = 0x00000000000000000000000000000000
ref1 = classical_encrypt(pt1, K1)
print("Test 1: all-zero plaintext and key")
print(f"expected: 0x{ref1:016X}")
check_equal("efficient raw", quantum_encrypt(pt1, K1), ref1)
check_equal("efficient Clifford+T", quantum_encrypt(pt1, K1, True), ref1)
check_equal("relative-phase Clifford+T", quantum_encrypt(pt1, K1, relative_phase=True), ref1)

# Test 2: non-zero plaintext and key.
pt2 = 0x1111111111111111
K2 = 0x0ABCDEFFEDC123456789BA9876543210
ref2 = classical_encrypt(pt2, K2)
print("\nTest 2: non-zero plaintext and key")
print(f"expected: 0x{ref2:016X}")
check_equal("efficient raw", quantum_encrypt(pt2, K2), ref2)
check_equal("efficient Clifford+T", quantum_encrypt(pt2, K2, True), ref2)
check_equal("relative-phase Clifford+T", quantum_encrypt(pt2, K2, relative_phase=True), ref2)

# Test 3: reversibility, encrypt then decrypt returns plaintext.
print("\nTest 3: reversibility")
ct2 = quantum_encrypt(pt2, K2)
check_equal("classical decrypt", classical_decrypt(ct2, K2), pt2)
check_equal("quantum decrypt raw", quantum_decrypt(ct2, K2), pt2)
check_equal("quantum decrypt Clifford+T", quantum_decrypt(ct2, K2, True), pt2)
check_equal("relative-phase decrypt", quantum_decrypt(ct2, K2, relative_phase=True), pt2)

# Test 4: alpha-reflection decryption path.
print("\nTest 4: alpha-reflection")
ct4 = classical_encrypt(pt2, K2)
rt4_classical = classical_decrypt(ct4, K2)
rt4_quantum = quantum_decrypt(ct4, K2)
check_equal("classical alpha-reflection", rt4_classical, pt2)
check_equal("quantum alpha-reflection", rt4_quantum, pt2)
check_equal("relative-phase alpha-reflection", quantum_decrypt(ct4, K2, relative_phase=True), pt2)
```
